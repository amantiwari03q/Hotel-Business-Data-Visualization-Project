```python
import pandas as pd

df = pd.read_csv(r"C:\Users\Aman Tiwari\Downloads\hotel_bookings_data.csv")

print(df.shape)
df.head()
```

### Data Overview

```python
df.info()
df.describe()
```

### Missing Values

```python
df.isnull().sum().sort_values(ascending=False)
```

```python
# Handle missing values

df['children'] = df['children'].fillna(0)
df['agent'] = df['agent'].fillna(0)
df['company'] = df['company'].fillna(0)
df['city'] = df['city'].fillna('Unknown')
```

```python
df.isnull().sum().sort_values(ascending=False)
```

### Duplicate Rows

```python
# Duplicate rows

print(df.duplicated().sum())

df = df.drop_duplicates()
```

### Anomaly Check

```python
df = df[
    (df['adults'] + df['children'] + df['babies']) > 0
]
```

```python
df = df[df['adr'] >= 0]
```

```python
df['total_stay'] = ( df['stays_in_weekend_nights'] + df['stays_in_weekdays_nights'] )
```

```python
df['total_guests'] = (
    df['adults']
    + df['children']
    + df['babies']
)
```

### Booking Share

```python
hotel_counts = df['hotel'].value_counts()

sns.countplot(data=df, x='hotel')
plt.title("Bookings by Hotel Type")
plt.xlabel("Hotel Type")
plt.ylabel("Bookings")
plt.show()
```

### Monthly Booking Analysis

```python
monthly = df.groupby(
    ['arrival_date_month','hotel']
).size().reset_index(name='bookings')
```

```python
#line chart

sns.lineplot(
    data=monthly,
    x='arrival_date_month',
    y='bookings',
    hue='hotel'
)
```

### Cancellation Rate

```python
cancel = df.groupby('hotel')['is_canceled'].mean()*100

cancel.plot(kind='bar')
plt.ylabel("Cancellation %")
plt.show()
```

### Stay Duration Analysis

```python
stay_cancel = df.groupby(
    'total_stay'
)['is_canceled'].mean()*100
```

```python
stay_cancel.plot()
plt.xlabel("Stay Nights")
plt.ylabel("Cancellation %")
```

### Lead Time Analysis

```python
lead = df.groupby(
    'lead_time'
)['is_canceled'].mean()*100
```

```python
lead.plot()
plt.xlabel("Lead Time")
plt.ylabel("Cancellation %")
```
